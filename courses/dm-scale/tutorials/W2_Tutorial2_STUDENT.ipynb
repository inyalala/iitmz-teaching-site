## Section A — Correlation Matrix (WORKED EXAMPLE, done for you)

```python
# This cell is complete -- run it and confirm no errors.
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
from scipy.spatial.distance import euclidean, cityblock, jaccard

df = pd.read_csv("datasets/W1_zanzibar_crop_sample.csv")
numeric_cols = ["farm_size_ha", "yield_tonnes", "rainfall_mm", "price_per_kg", "revenue_usd"]
corr_matrix = df[numeric_cols].corr()
corr_matrix.round(3)
```

```python
# Also complete: visualize the correlation matrix as a heatmap
fig, ax = plt.subplots(figsize=(6,5))
im = ax.imshow(corr_matrix, cmap="Greys", vmin=-1, vmax=1)
ax.set_xticks(range(len(numeric_cols))); ax.set_xticklabels(numeric_cols, rotation=45, ha="right")
ax.set_yticks(range(len(numeric_cols))); ax.set_yticklabels(numeric_cols)
for i in range(len(numeric_cols)):
    for j in range(len(numeric_cols)):
        ax.text(j, i, f"{corr_matrix.iloc[i,j]:.2f}", ha="center", va="center", fontsize=8)
plt.colorbar(im)
plt.title("Correlation Matrix -- Zanzibar Crop Dataset")
plt.tight_layout()
plt.show()
```
