## 1. Insertion Sort & Inversion Sensitivity

Insertion sort performs one shift per inversion, running in $O(n + I)$ where $I$ is the number of inverted pairs.

```python
def insertion_sort(arr):
    comps = shifts = 0
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0:
            comps += 1
            if arr[j] > key:
                arr[j + 1] = arr[j]
                shifts += 1
                j -= 1
            else:
                break
        arr[j + 1] = key
    return comps, shifts


a = [4, 2, 5, 1, 3]
comps, shifts = insertion_sort(a)
assert a == [1, 2, 3, 4, 5] and shifts == 6
print(f"Sorted: {a}, comparisons={comps}, shifts={shifts}")
```

## 2. `ArrayStack`, `LinkedStack`, Bracket Matching & Postfix Evaluation

Both stack implementations satisfy the exact same LIFO ADT contract so client algorithms (`is_balanced_brackets`, `eval_postfix`, `MinStack`) run interchangeably on either representation.

```python
class ArrayStack:
    def __init__(self, cap=4):
        self._data = [None] * cap
        self._top = -1

    def is_empty(self):
        return self._top == -1

    def __len__(self):
        return self._top + 1

    def push(self, x):
        if self._top + 1 == len(self._data):
            nb = [None] * (2 * len(self._data))
            for i in range(self._top + 1):
                nb[i] = self._data[i]
            self._data = nb
        self._top += 1
        self._data[self._top] = x

    def pop(self):
        if self._top == -1:
            raise IndexError("empty")
        v = self._data[self._top]
        self._data[self._top] = None
        self._top -= 1
        return v

    def peek(self):
        if self._top == -1:
            raise IndexError("empty")
        return self._data[self._top]


class Node:
    def __init__(self, data, nxt=None):
        self.data, self.next = data, nxt


class LinkedStack:
    def __init__(self):
        self._head, self._size = None, 0

    def is_empty(self):
        return self._head is None

    def __len__(self):
        return self._size

    def push(self, x):
        self._head = Node(x, self._head)
        self._size += 1

    def pop(self):
        if self._head is None:
            raise IndexError("empty")
        v = self._head.data
        self._head = self._head.next
        self._size -= 1
        return v

    def peek(self):
        if self._head is None:
            raise IndexError("empty")
        return self._head.data


def eval_postfix(expr):
    s = LinkedStack()
    for tok in expr.split():
        if tok in "+-*/":
            b, a = s.pop(), s.pop()
            s.push(a + b if tok == "+" else a - b if tok == "-" else a * b if tok == "*" else a / b)
        else:
            s.push(float(tok))
    return s.pop()


assert eval_postfix("5 1 2 + 4 * + 3 -") == 14.0
print("Postfix '5 1 2 + 4 * + 3 -' =", eval_postfix("5 1 2 + 4 * + 3 -"))
```
