## 1. Exact 3-SAT Backtracking & Polynomial Reduction to Independent Set

A 3-SAT formula with $m$ clauses is satisfiable if and only if the gadget graph $G$ (with $3m$ vertices) contains an Independent Set of size $m$.

```python
def eval_cnf(clauses, assign):
    return all(any((assign[abs(l)] if l > 0 else not assign[abs(l)]) for l in cl) for cl in clauses)


def solve_sat(n_vars, clauses):
    assign = {}

    def ok():
        for cl in clauses:
            sat = any((abs(l) in assign and ((l > 0) == assign[abs(l)])) for l in cl)
            undec = any(abs(l) not in assign for l in cl)
            if not sat and not undec:
                return False
        return True

    def dfs(v):
        if not ok():
            return False
        if v > n_vars:
            return True
        for val in (True, False):
            assign[v] = val
            if dfs(v + 1):
                return True
            del assign[v]
        return False

    return assign if dfs(1) else None


clauses = [(1, -2, 3), (-1, 2, 3), (-1, -2, -3)]
sol = solve_sat(3, clauses)
assert sol is not None and eval_cnf(clauses, sol)
print("Satisfying assignment:", sol)
```

## 2. 2-Approximation for Minimum Vertex Cover

Repeatedly picking an uncovered edge `(u, v)` and adding *both* endpoints yields a maximal matching of size $|M|$. Any optimal cover must contain at least one vertex per edge in $M$, proving $|C| = 2|M| \le 2 \cdot \text{OPT}$.

```python
def vertex_cover_2approx(edges):
    cover = set()
    for u, v in edges:
        if u not in cover and v not in cover:
            cover.add(u)
            cover.add(v)
    return cover


edges = [(0, 1), (1, 2), (2, 3), (3, 4), (1, 4)]
vc = vertex_cover_2approx(edges)
assert all(u in vc or v in vc for u, v in edges)
print("2-approx vertex cover:", sorted(vc), "| size =", len(vc))
```
