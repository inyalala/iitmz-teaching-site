## 1. Merge Sort & $O(n \log n)$ Inversion Counting

Whenever `right[j] < left[i]` during the merge step, `right[j]` forms an inversion with every remaining item `left[i:]`, adding `len(left) - i` inversions in $O(1)$.

```python
import random


def merge_sort_and_count(arr):
    if len(arr) <= 1:
        return arr[:], 0
    mid = len(arr) // 2
    left, il = merge_sort_and_count(arr[:mid])
    right, ir = merge_sort_and_count(arr[mid:])
    out, i, j, invs = [], 0, 0, il + ir
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            out.append(left[i]); i += 1
        else:
            out.append(right[j])
            invs += len(left) - i
            j += 1
    return out + left[i:] + right[j:], invs


s, invs = merge_sort_and_count([4, 2, 5, 1, 3])
assert s == [1, 2, 3, 4, 5] and invs == 6
print("Sorted:", s, "| Inversions:", invs)
```

## 2. Randomized Quicksort & Expected-$O(n)$ Quickselect

Random pivot choice turns sorted-input worst-case $O(n^2)$ into expected $O(n \log n)$ for Quicksort and expected $O(n)$ ($T(n) = T(n/2) + \Theta(n)$) for Quickselect.

```python
def quickselect(arr, k, seed=42):
    a = arr[:]
    rng = random.Random(seed)
    lo, hi = 0, len(a) - 1
    while lo < hi:
        p = rng.randint(lo, hi)
        a[p], a[hi] = a[hi], a[p]
        pivot, store = a[hi], lo
        for i in range(lo, hi):
            if a[i] < pivot:
                a[store], a[i] = a[i], a[store]
                store += 1
        a[store], a[hi] = a[hi], a[store]
        if store == k:
            return a[store]
        if k < store:
            hi = store - 1
        else:
            lo = store + 1
    return a[lo]


rng = random.Random(10)
vals = [rng.randint(1, 500) for _ in range(99)]
med = quickselect(vals, len(vals) // 2)
assert med == sorted(vals)[len(vals) // 2]
print("Quickselect median:", med)
```
