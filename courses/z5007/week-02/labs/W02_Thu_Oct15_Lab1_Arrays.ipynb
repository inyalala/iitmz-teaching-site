## Warm-up (3:38–3:42): Halving finds a number fast

The computer picks a secret number from 1 to 1000, then finds it by always guessing **the middle** of what is still possible and throwing away the wrong half.

**Just run the cell below** (Shift + Enter) and read the lines it prints. Nothing to type. Count the guesses. Run it again: a new secret, but never more than 10 guesses.

````python
import random

class Game:
    def __init__(self, low=1, high=1000):
        self.secret = random.randint(low, high)
        self.tries = 0
        print(f"I picked a secret number from {low} to {high}. Make your first guess in the next cell.")

    def guess(self, g):
        self.tries += 1
        if g == self.secret:
            print(f"guess {self.tries}: {g}  CORRECT! You found it in {self.tries} guesses.")
        else:
            hint = "too LOW: the secret is BIGGER, try a larger number" if g < self.secret else "too HIGH: the secret is SMALLER, try a smaller number"
            print(f"guess {self.tries}: {g} is {hint}")

game = Game()      # run this cell once
````

````python
game.guess(400)    # change the number, run this cell again (Ctrl + Enter) until you win
````

````python
def computer_play(low=1, high=1000):
    secret = random.randint(low, high)
    lo, hi, tries = low, high, 0
    while True:
        mid = (lo + hi) // 2               # always guess the middle of what is left
        tries += 1
        print(f"guess {tries}: {mid}   (still possible: {lo}..{hi}, {hi - lo + 1} numbers)")
        if mid == secret:
            print(f"Computer found {secret} in {tries} guesses."); return tries
        if mid < secret: lo = mid + 1
        else: hi = mid - 1

computer_play()
````

## Step 4 (4:20–4:38). The dynamic array

**Goal:** build the dynamic array from this morning's lecture, line by line.

**The idea.** Two numbers: `size` (items stored) and `cap` (boxes reserved).
* `append(x)`: if `size == cap` the block is full, so first **resize** to a bigger block. Then write x into box `size` and add 1 to `size`.
* `_resize(new_cap)`: make a new block of `new_cap` boxes, **copy** every item across (count each copy in `self.copies`), then use the new block.
* `pop()`: remove and return the last item. If the array is now only **one quarter full**, shrink to half the capacity.

How much to grow is decided by `self._bigger()`, which is **given**: by default it doubles. In Step 5 we change it to compare growth rules.

```

new block:     [  ][  ][  ][  ][  ][  ][  ][  ]
copy 4 items:  [12][ 7][33][ 5][  ][  ][  ][  ]      copies += 4
write 41:      [12][ 7][33][ 5][41][  ][  ][  ]      size = 5, cap = 8
```
