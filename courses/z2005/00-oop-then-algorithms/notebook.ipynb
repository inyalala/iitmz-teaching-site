## 1. A class is a blueprint

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def distance_to(self, other):
        return ((self.x - other.x) ** 2 + (self.y - other.y) ** 2) ** 0.5

p1 = Point(0, 0)
p2 = Point(3, 4)
print(p1.distance_to(p2))  # expect 5.0
```

## 2. Encapsulation

`withdraw` refuses to take the balance negative. This is what the A0
assignment asks students to add themselves; here it is already in place.

```python
class BankAccount:
    def __init__(self, balance=0):
        self._balance = balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("deposit must be positive")
        self._balance += amount

    def withdraw(self, amount):
        if amount > self._balance:
            raise ValueError("insufficient funds")
        self._balance -= amount

    def balance(self):
        return self._balance
```

```python
acct = BankAccount(100)
acct.withdraw(30)
assert acct.balance() == 70
try:
    acct.withdraw(1000)
    raise AssertionError("withdraw should have raised ValueError")
except ValueError:
    print("withdraw correctly rejected an overdraw")
```

## 3. Inheritance and polymorphism

`Triangle` is what students add in A0. Solved here so the loop below can be
demonstrated live with all three shapes.

```python
class Shape:
    def area(self):
        raise NotImplementedError

class Circle(Shape):
    def __init__(self, r):
        self.r = r
    def area(self):
        return 3.14159 * self.r ** 2

class Square(Shape):
    def __init__(self, s):
        self.s = s
    def area(self):
        return self.s ** 2

class Triangle(Shape):
    def __init__(self, b, h):
        self.b = b
        self.h = h
    def area(self):
        return 0.5 * self.b * self.h

for shape in [Circle(2), Square(3), Triangle(4, 5)]:
    print(type(shape).__name__, shape.area())
```

## 4. Why data structures are classes

This `Stack` is the pattern every structure in this course follows: state
hidden in `__init__`, behavior exposed as methods.

```python
class Stack:
    def __init__(self):
        self._items = []

    def push(self, item):
        self._items.append(item)

    def pop(self):
        return self._items.pop()

    def is_empty(self):
        return len(self._items) == 0

s = Stack()
s.push(1)
s.push(2)
assert s.pop() == 2
assert not s.is_empty()
print("stack behaves as expected")
```

## 5. MTech addition — composition vs. inheritance

```python
class Engine:
    def start(self):
        return "engine started"

class Car:
    def __init__(self):
        self.engine = Engine()   # composition: Car HAS an Engine, is not one

    def start(self):
        return self.engine.start()

print(Car().start())
```

**Discussion point for MTech:** `Car` is not a kind of `Engine`, so
inheriting from it would be wrong even though it would technically work.
Ask a Car-inherits-Engine student to explain what `car.start()` even means
if `Car` *is an* `Engine` — the metaphor breaks immediately.
